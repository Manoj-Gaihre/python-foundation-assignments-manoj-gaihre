that reads the CSV using `csv.DictReader` and returns the **total value** of the

**Requirements**

````python
# --- SETUP: run this cell first, do not modify ---
import csv

with open("products.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["name", "price", "quantity"])
    writer.writerow(["Notebook", "2.50", 20]) # for checking the round actually works or not simply change the value of price of notebook to 2.555
    writer.writerow(["Pen", "1.00", 50])
    writer.writerow(["Backpack", "35.00", 3])

print("products.csv created.")
````

````python
# --- YOUR ANSWER ---
import csv

def total_inventory_value(path):
    total_value = 0
    with open("products.csv", "r", newline="") as r:
        reader = csv.DictReader(r)
        for row in reader:
            # print(row["price"])
            # print(row["quantity"])
            # print(type(row["price"]))
            # print(type(row["quantity"]))
            # print(row)

            price = float(row["price"])
            # print(type(price))

            quantity = int(row["quantity"])
            # print(type(quantity))


            total = price *quantity

            total_value += total

    return round(total_value, 2)
    # return total_value

            


# --- Test your function here ---
print(total_inventory_value("products.csv"))   # expected: 205.0
````
