# Exercise 3: Clean Numeric Values

### Given:

raw_values = [100, None, 250, "invalid", 300, None, 450]

### Create a new list containing only valid integers.

### First solve it using:

A loop

continue

isinstance()

### Expected result:

[100, 250, 300, 450]

### Then solve it again using a list comprehension.

```python
# using loop continue and isinstance()
raw_values = [100, None, 250, "invalid", 300, None, 450]
cleaned_values = []

for value in raw_values:
    if not isinstance(value, int):
        continue
    else:
        cleaned_values.append(value)


print(cleaned_values)
```

```python
# using list comprehension
raw_values = [100, None, 250, "invalid", 300, None, 450]
cleaned_values = []


for value in raw_values:
    cleaned_values = [value for value in raw_values if isinstance(value, int)]

print(cleaned_values)
```
