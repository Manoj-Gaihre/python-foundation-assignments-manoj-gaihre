# Stretch Exercise: Contact Book Menu

### Build an interactive contact book using a while loop.

### Display this menu:

1. Add contact

2. Search contact

3. Delete contact

4. Display all contacts

5. Exit

### Store contacts in a dictionary.

### Each contact should contain:

Name

Phone number

Email address

### Requirements:

Do not crash when searching for a missing contact.

Do not crash when deleting a missing contact.

Continue showing the menu until the user selects Exit.

Use break to stop the program.

Use nested dictionaries for the contact details.


```python
contact = {}

while True:
    print("-----Menu-----")
    print("1. Add contact")
    print("2. Search contact")
    print("3. Delete contact")
    print("4. Display all contact")
    print("5. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        print("Adding contact")
        name = input("Enter your name: ")
        phone_number = input("Enter your phone number: ")
        email = input("Enter your email: ")
        contact[name] = {
            "phone" : phone_number,
            "email" : email
        }
        # print(contact)
        print("Contact added successfully")
    elif choice == 2:
        print("Searching contact")
        searched_name = input("Enter your name to be searched: ")
        if searched_name in contact:
            print("user found")
            print(f"Name: {searched_name}")
            print(f"Phone Number: {contact[searched_name]['phone']}")
            print(f"Email: {contact[searched_name]['email']}")
        else:
            print("User not found. Please enter the valid name")
    elif choice == 3:
        print("Deleting contact")
        deleting_name = input("Enter the name to be deleted: ")
        if deleting_name in contact:
            del contact[deleting_name]
            print(f"User {deleting_name} deleted successfully.")
        else:
            print(f"Enter valid user name. {deleting_name} cannot be found!!!")
    elif choice == 4:
        print("Displaying all contact")
        if contact:
            for key, value in contact.items():
                print(f"name: {key}")
                print(f"Phone Number: {value['phone']}")
                print(f"Email Address: {value['email']}")
        else:
            print("No users added yet!!!")
    elif choice == 5:
        print("Existing.....")
        break
    else:
        print("Choice the valid number!!!!")
```
