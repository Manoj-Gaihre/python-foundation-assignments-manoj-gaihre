# Exercise 7: Nested Order Summary

### Given:

orders = {

    "ORD-001": {

        "customer": "Anisha",

        "amount": 2500,

        "status": "Completed"

    },

    "ORD-002": {

        "customer": "Ravi",

        "amount": 1800,

        "status": "Pending"

    },

    "ORD-003": {

        "customer": "Maya",

        "amount": 3200,

        "status": "Completed"

    }

}

### Tasks:

Print every order ID and customer.

Print only completed orders.

Calculate the total amount of completed orders.

Count pending orders.

Add a new order to the dictionary.



```python
orders = {
    "ORD-001": {
        "customer": "Anisha",
        "amount": 2500,
        "status": "Completed"
    },
    "ORD-002": {
        "customer": "Ravi",
        "amount": 1800,
        "status": "Pending"
    },
    "ORD-003": {
        "customer": "Maya",
        "amount": 3200,
        "status": "Completed"
    }
}
```

```python
# Print every order ID and customer.
for order_id, order_details in orders.items():
    print(f"{order_id} : {order_details['customer']}")
```

```python
# Print only completed orders.
for order_id, order_details in orders.items():
    if order_details['status'] == "Completed":
        print(order_details)
```

```python
# Calculate the total amount of completed orders.
total_amount = 0
for order_id, order_details in orders.items():
    if order_details['status'] == "Completed":
        total_amount += order_details['amount']

print(total_amount)
```

```python
# Count pending orders.
pending_orders = 0
for order_id, order_details in orders.items():
    if order_details['status'] == "Pending":
        pending_orders = pending_orders+1

print(pending_orders)
```

```python
# Add a new order to the dictionary.
orders.update({"ORD-004":{
    "customer": "Manoj",
    "amount": 3000,
    "status": "Pending"
}})
print(orders)
```
