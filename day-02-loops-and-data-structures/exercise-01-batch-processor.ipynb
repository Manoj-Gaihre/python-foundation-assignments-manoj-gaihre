# Exercise 1: Batch Processor

### Use a for loop and range() to print batch numbers from 1 to 10.

### After every third batch, display:

Checkpoint reached

### Example:

Processing batch 1
Processing batch 2
Processing batch 3
Checkpoint reached

### Hint:

### Use the modulo operator.

batch_number % 3 == 0

```python
for i in range(1, 11):
    if i % 3 == 0:
        print(f"Processing batch {i}")
        print(f"Checkpoint reached")
    else:
        print(f"Processing batch {i}")
```
