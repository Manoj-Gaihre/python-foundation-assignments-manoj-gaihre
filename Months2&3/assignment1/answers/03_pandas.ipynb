### C12 — Aggregation and a summary table (3 pts)
On the cleaned data, produce: 

(a) mean, max and count of `marks` per subject using `groupby().agg()`;

(b) mean `marks` by `city` and `subject`; 

(c) a `pivot_table` of mean marks with cities as rows and subjects as columns; 

(d) mean marks per `exam_month` , sorted chronologically. Export (a) to `outputs/summary.csv`, and state in one sentence what the month trend shows.

```python
import pandas as pd

# Load the raw data
df = pd.read_csv("../data/scores_raw.csv")

# cleaning the data set

df["city"] = df["city"].str.strip().str.title()

df["attendance_percent"] = df["attendance_percent"].fillna(df["attendance_percent"].mean())

df = df[~df["marks"].isna()]

df = df.drop_duplicates()

df["exam_date"] = pd.to_datetime(df["exam_date"])

df["exam_month"] = df["exam_date"].dt.month
df["exam_weekday"] = df["exam_date"].dt.weekday

renamed_df = df.rename(columns= {"attendance_percent":"attendance"})

df.loc[df["marks"] > 100, "marks"] = np.nan

df["passed"] = df["marks"] >= 40

df["grade"] = np.where(
    df["marks"] >= 85, "A", np.where(
        df["marks"] >= 70, "B", np.where(
            df["marks"] >= 40, "C", "F"
        )
    )
)

df["subject"] = df["subject"].astype("category")
df["city"] = df["city"].astype("category")



df.info()
```

```python
# (a) mean, max and count of `marks` per subject using `groupby().agg()`;

subject_summary = df.groupby("subject")["marks"].agg([
    "mean",
    "max",
    "count"
])
print("C12a - aggs: ")
print(subject_summary)
```

```python
# (b) mean `marks` by `city` and `subject`;

print("C12b - mean by city and subject: ")
df.groupby(["city","subject"])["marks"].mean()
```

```python
# (c) a `pivot_table` of mean marks with cities as rows and subjects as columns;

print("C12c - pivot table: ")
df.pivot_table(
    values= "marks",
    index = "city",
    columns = "subject",
    aggfunc = "mean"
)
```

```python
# (d) mean marks per `exam_month` , sorted chronologically.

print("C12d - mean per exam month sorted chronologically: ")
df.groupby("exam_month")["marks"].mean().sort_index()
```

```python
# Export (a) to `outputs/summary.csv`, and state in one sentence what the month trend shows.

subject_summary.to_csv("../outputs/summary.csv")

print("C12 - Summary exported to outputs/summary.csv")
```

### A reporting pipeline. Wrap Part C into three functions — `load(path)`, `clean(df)`, `summarise(df)` — so that `summarise(clean(load("data/scores_raw.csv")))` reproduces your C12(a) table from the raw file in one line. It must run top to bottom with no manual steps.

```python
import pandas as pd
import numpy as np

def load(path):
    return pd.read_csv(path)

def clean(df):
    df = df.copy()
    df["exam_date"] = pd.to_datetime(df["exam_date"])

    # cleaning the data set

    df["city"] = df["city"].str.strip().str.title()

    df["attendance_percent"] = df["attendance_percent"].fillna(df["attendance_percent"].mean())

    df = df[~df["marks"].isna()]

    df = df.drop_duplicates()

    df["exam_date"] = pd.to_datetime(df["exam_date"])

    df["exam_month"] = df["exam_date"].dt.month
    df["exam_weekday"] = df["exam_date"].dt.weekday

    renamed_df = df.rename(columns= {"attendance_percent":"attendance"})

    df.loc[df["marks"] > 100, "marks"] = np.nan

    df["passed"] = df["marks"] >= 40

    df["grade"] = np.where(
        df["marks"] >= 85, "A", np.where(
            df["marks"] >= 70, "B", np.where(
                df["marks"] >= 40, "C", "F"
            )
        )
    )

    df["subject"] = df["subject"].astype("category")
    df["city"] = df["city"].astype("category")

    return df

def summarise(df):
    return df.groupby("subject")["marks"].agg([
        "mean",
        "max",
        "count"
    ])


# running the code

summary = summarise(clean(load("../data/scores_raw.csv")))
print(summary)
```

### Where your cleaning could mislead. 150–250 words in `answers/03_pandas.ipynb`: pick two cleaning decisions you made (mean-filling attendance, dropping missing marks, deleting duplicates, whatever you did with the impossible marks) and explain, concretely, how each one could bend the conclusions in your C12 summary.

Two cleaning decisions I made could mislead the conclusion of the C12 summary. 

First, I filled missing `attendance_percent` values with the overall mean attendance. It assumes that students with missing attendance had attendance similar to the mean attendance. But if the missing attendence could belong to the student with high or low attendence, then this decision of filling with mean could be inappropraite as mean filling could reduce the variation and also makes the relationship between attendance and marks appear weaker or more typical than it really was.

Second, I handled missing and the impossible values in the `marks` column by removing rows with missing marks and filling the value with `NaN`. This changes the total number of the rows and count of the records. It can also change the mean if the removed students would have marks that were high or low than the other students.
