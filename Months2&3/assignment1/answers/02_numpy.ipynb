### B7 — Matrix multiplication (4 pts)
(a) On the grocery-basket example (`quantities = [2, 3, 1]`, `prices = [120, 90, 250]`), print both `*` and `@`, and state in a comment what each returned and why they differ. 

(b) Build `features` of shape `(14,2)` whose columns are the across-city NaN-aware mean temperature and mean humidity per day, define `weights = np.array([0.6, -0.1])`, and compute a comfort score per day with `@`. Print the shapes of `features`, `weights`, and the result, and explain in a comment why the inner dimensions had to match.

```python
quantities = np.array([2, 3, 1])
prices = np.array([120, 90, 250])

print("B7a - *: ", quantities * prices)
print("B7a - @: ", quantities @ prices)

# * performs element-wise multiplication:
# [2*120, 3*90, 1*250] = [240, 270, 250]
#
# @ performs dot product and gives the total grocery cost:
# (2*120) + (3*90) + (1*250) = 760
```

```python
mean_temperature = np.nanmean(temps, axis = 0)
mean_humidity = np.nanmean(humidity, axis = 0)
features = np.array([mean_temperature, mean_humidity]).T
print("B7b - Shape of features is: ",features.shape)

weights = np.array([0.6, -0.1])
print("B7b - Shape of weights is: ",weights.shape)

result = features @ weights
print("B7b - Comfort score per day with @: ", result)
print("B7b - Shape of result is: ",result.shape)

# Matrix Multiplication requires the inner dimensions to match. features has shape (14,2) and weights has shape (2,), 
# so the inner dimension 2 matches and the result has shape (14,).
```

### B9 — Reproducible randomness and a train/test split (3 pts)
Flatten `temps` to 70 samples. With `np.random.seed(42)`, shuffle an index array and split it 80/20 into train and test index sets; print both shapes, confirm they sum to 70, and confirm they share no index. Re-seed with 42, repeat, and assert the split is identical. In a comment, explain why you shuffle before splitting, and why the seed matters when you share results.

```python
flattened = temps.flatten()
print(flattened)

index = np.arange(flattened.size)
np.random.seed(42)
np.random.shuffle(index)

# split
split_point = int(0.8 * len(index))

train_idx = index[:split_point]
test_idx = index[split_point:]

print("B9 - Train index shape: ",train_idx.shape)
print("B9 - Test index shape: ", test_idx.shape)

print("B9 - confirm shape sum to 70: ", len(train_idx) + len(test_idx))

# train and test has no common indices
print("B9 - Share no index: ",len(np.intersect1d(train_idx, test_idx)))


# Repeat the exact same process with the same seed
index2 = np.arange(flattened.size)

np.random.seed(42)  # for conforming simply comment and uncomment this line 

np.random.shuffle(index2)

# split
split_point2 = int(0.8 * len(index2))

train_idx2 = index2[:split_point2]
test_idx2 = index2[split_point2:]

print("\nB9 - Repeated Train index shape: ",train_idx2.shape)
print("B9 - Repeated Test index shape: ", test_idx2.shape)

assert np.array_equal(train_idx, train_idx2)
assert np.array_equal(test_idx, test_idx2)

print("Repeated split is identical: True")

# We shuffle before splitting so that the train and test sets are randomly selected rather than simply taking the first 80% and 
# last 20% of the data.
# 
# The seed matters because it makes the random shuffle reproducible. Anyone using the same data and seed can reproduce the same split.
```

```python
label_grid = np.where(temps > 30, "hot",
                      np.where(temps >= 22, "mild", "cold")
                    )
print("B10a - Label grid: ", label_grid)
print("B10a - Label grid shape: ", label_grid.shape)
```

```python
scores = np.array([0.08, 0.62, 0.21, 0.09])
labels = ["rain", "clear", "cloudy", "fog"]

index = np.argmax(scores)
label = labels[index]

print("B10b - Predicted label: ", label)

# max() gives the largest score but a classifier needs to know which class has that score for that argmax() gives the index 
# of the largest score not the largest value.
```
