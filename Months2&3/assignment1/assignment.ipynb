# Remotes and the everyday workflow

### Explain the relationship between your local repository and `origin` . What do `push` and `pull` each move, in which direction, and why is pulling before pushing the habit to build?

#### Answer
Local repository is the Git repository on computer, while origin is the name git gives to the remote repository from which we cloned.

`git push` moves from local repository to the remote repository.

`git pull` moves from remote repository to local repository.

Pulling is done before pushing to make sure our local branch includes the latest changes from the remote repository. Suppose a person is working in a team of 5 members, and another person has already pushed changes, then if he also pushed the same changes then there could be conflict.

So, in order to avoid push rejection and to reduce the chance of conflicts, pulling is done before pushing.

### `git fetch` and `git pull` are not the same command. What is the difference, and when would you deliberately choose fetch ?

#### Answer:

`git fetch` downloads the latest changes from the remote repository but does not merge them into current branch. It allows us to inspect the changes before deciding what to do.

`git pull` downloads the latest changes and automatically integrates them into current branch.

I use fetch when I want to inspect what changed on GitHub before deciding whether or how to integrate it.

# Branching, merging, pull requests

### A branch in Git is often described as "just a pointer." Explain what that means, and explain concretely what goes wrong on a team when everyone commits directly to `main`.

#### Answer:

A branch in git is just a pointer to a specific commit. It is a movable label that keeps track of the latest commit in a particular line of development.

If everyone works directly on `main`, multiple developers can make changes to the same branch at the same time. This can cause:
- Merge conflicts
- unstable code
- difficult reviews

Branches allow people to work separately and then review changes through a controlled process.

### A merge conflict happens when two branches change the same lines of the same file. Explain why Git cannot resolve this automatically, what the `<<<<<<<`, `=======`, `>>>>>>>` markers mean, and what you must do to finish the merge.

#### Answer:
A merge conflict happens when two branches change the same lines of the same file. Git cannot resolve it automatically because git does not know the developer's intended result and both changes may be valid.

`<<<<<<<` it means current branch's version

`=======` it means boundary between the two versions

`>>>>>>>` it means other branch's version

What you must do to finish the merge?
- Open the conflicted file.
- Review both versions of the code.
- Decide which change to keep, or combine both changes correctly.
- Remove the conflict markers(<<<<<<<, =======, >>>>>>>).
- Save the file.
- Stage the resolved file:
    `git add filename`
- Complete the merge:
    `git commit`
    
git will then record the merge as completed.

### Why should `.gitignore` be one of your first commits? If a file is already tracked, does adding it to `.gitignore` stop Git from tracking it — and if not, what do you do instead?

#### Answer:
.gitignore should be one of our first commits because it tells Git which files and folders should not be tracked in the repository. This helps prevent accidentally committing unnecessary or sensitive files.

If a file is already tracked, adding it to .gitignore doesnot stop git from tracking it.

### Explain the difference between `.env` and `.env.example`, and why they get opposite treatment. If a real API key was committed three weeks ago, why is deleting it in a new commit not a fix, and what should actually be done?

`.env` contains real secret/local configuration. 

For example:

`API_KEY=actual_secret_key`

`DATABASE_PASSWORD=actual_password`

It should normally be ignored.

`.env.example` contains the expected variable names but not the actual secrets.

This file should normally be committed so other developers know what configuration they need.

If a real API key was committed, simply deleting it in a later commit isn't enough because the secret remains in Git history.

The correct response is to revoke/rotate the exposed key immediately, then remove the secret from the repository/history as appropriate.

### What problem do a virtual environment and `requirements.txt` solve together? Why is `venv/` never committed, when  `requirements.txt` always is?  

#### Answer:
A virtual environment and requeirements.txt solve the problem of dependency management and reproducibility.
- A virtual environment creates an isolated Python environment for the project, so its packages do not interfere with other projects or the system Python installation.
- requirements.txt records the Python packages and versions that the project needs, so another developer can install the same dependencies.

For example:

`python -m venv venv`

Then install the required packages:

`pip install pandas numpy`

Save them to:

`pip install -r requirements.txt`

Someone else can then install:

`pip install -r requirements.txt`

Why is `venv/` never committed, when  `requirements.txt` always is?

The virtual environment folder(venv/) contains installed packages, executables, and environment-specific files. It can be very large and may differ between operating systems and machines, so it should not be committed.

The requirements.txt file is a small text file that describes what packages the project needs. It should be committed so other developers can recreate the environment.
