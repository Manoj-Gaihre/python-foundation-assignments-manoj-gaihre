```python
# Creating the table
cursor.execute("""
    CREATE TABLE IF NOT EXISTS employees (
        employee_id INT primary key,
        employee_code VARCHAR(50),
        first_name VARCHAR(100),
        last_name VARCHAR(100),
        gender VARCHAR(20),
        age INT,
        city VARCHAR(100),
        department VARCHAR(100),
        job_title VARCHAR(100),
        employment_type VARCHAR(50),
        join_date DATE,
        years_experience INT,
        monthly_salary NUMERIC(12,2),
        annual_bonus NUMERIC(12,2),
        performance_rating NUMERIC(3,2),
        projects_completed INT,
        leave_days_taken INT,
        overtime_hours NUMERIC(8,2),
        remote_worker BOOLEAN,
        employment_status VARCHAR(50),
        education_level VARCHAR(100),
        email VARCHAR(150),
        phone VARCHAR(30),
        emergency_contact VARCHAR(30),
        manager_name VARCHAR(100),
        certification VARCHAR(150),
        work_shift VARCHAR(50),
        performance_category VARCHAR(50),
        promotion_eligible BOOLEAN
    )
""")

conn.commit()
```

## Arithmatic Operator

```python
# 1. Display employee_id, monthly_salary, and monthly_salary + 10000 as increased_salary.
cursor.execute("select employee_id, monthly_salary, monthly_salary + 10000 as increased_salary from employees")
rows = cursor.fetchall()
print(rows)
```

```python
# 2. Calculate reduced_salary by subtracting 5000 from monthly_salary.
cursor.execute("select employee_id, monthly_salary, monthly_salary-5000 as reduced_salary from employees")
rows = cursor.fetchall()
print(rows)
```

```python
# 3. Calculate annual_salary by multiplying monthly_salary by 12.
cursor.execute("select monthly_salary * 12 as annual_salary from employees")
rows = cursor.fetchall()
print(rows)
# for row in rows:
#     print(row)
```

```python
# 4. Calculate average_monthly_bonus by dividing annual_bonus by 12.
cursor.execute("select annual_bonus / 12 as average_monthly_bonus from employees")
row = cursor.fetchall()
print(row)
```

```python
# 5. Display age and age % 2 as remainder.
cursor.execute("select age, age%2 as remaninder from employees")
row = cursor.fetchall()
print(row)
```

```python
# 6. Calculate total_compensation as (monthly_salary * 12) + annual_bonus.
cursor.execute("select (monthly_salary * 12) + annual_bonus as total_compensation from employees")
row = cursor.fetchall()
print(row)
```

## Comparison Operator

```python
# 7. Find employees whose monthly_salary is greater than 100000.
cursor.execute("select employee_id, monthly_salary from employees where monthly_salary > 100000")
row = cursor.fetchall()
print(row)
```

```python
# 8. Find employees whose age is less than 30.
cursor.execute("select employee_id,age from employees where age < 30")
row = cursor.fetchall()
print(row)
```

```python
# 9. Find employees with performance_rating greater than or equal to 4.5.
cursor.execute("select employee_id, performance_rating from employees where performance_rating >= 4.5")
row = cursor.fetchall()
print(row)
```

```python
# 10. Find employees whose years_experience is less than or equal to 5.
cursor.execute("select employee_id, years_experience from employees where years_experience <= 5")
row = cursor.fetchall()
print(row)
```

```python
# 11. Find employees whose department is equal to IT.
cursor.execute("select employee_id, department from employees where department = 'IT'")
row = cursor.fetchall()
print(row)
```

```python
# 12. Find employees whose employment_status is not equal to Active.
cursor.execute("select employee_id, employment_status from employees where employment_status <> 'Active'")
row = cursor.fetchall()
print(row)
```

## Logical Operator

```python
# 13. Find employees from Kathmandu AND monthly_salary greater than 100000.
cursor.execute("select employee_id, city, monthly_salary from employees where city = 'Kathmandu' and monthly_salary > 100000")
row = cursor.fetchall()
print(row)
```

```python
# 14. Find employees from Kathmandu OR Pokhara.
cursor.execute("select employee_id, city from employees where city = 'Kathmandu' or city = 'Pokhara'")
row = cursor.fetchall()
print(row)
```

```python
# 15. Find employees from IT OR Analytics with performance_rating greater than 4.
cursor.execute("select employee_id, department, performance_rating from employees where (department = 'IT' or department = 'Analytics') and performance_rating > 4")
row = cursor.fetchall()
print(row)
```

```python
# 16. Find employees who are NOT remote workers.
cursor.execute("select employee_id, remote_worker from employees where remote_worker = 'No'")
row = cursor.fetchall()
print(row)
```

```python
# 17. Find employees with age below 40 AND years_experience above 5 AND employment_status = Active.
cursor.execute("select employee_id, age, years_experience, employment_status from employees where age < 40 and years_experience > 5 and employment_status = 'Active'")
row = cursor.fetchall()
print(row)
```

```python
# 18. Find Full-Time employees with salary above 80000 OR performance_rating above 4.5.
cursor.execute("select employee_id, employment_type, monthly_salary, performance_rating from employees where employment_type = 'Full-Time' and (monthly_salary > 80000 or performance_rating > 4.5)")
row = cursor.fetchall()
print(row)
```
