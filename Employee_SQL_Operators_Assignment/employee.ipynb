```python
# Creating the table
cursor.execute("""
    CREATE TABLE IF NOT EXISTS employees (
        employee_id INT primary key,
        employee_code VARCHAR(50),
        first_name VARCHAR(100),
        last_name VARCHAR(100),
        gender VARCHAR(20),
        age INT,
        city VARCHAR(100),
        department VARCHAR(100),
        job_title VARCHAR(100),
        employment_type VARCHAR(50),
        join_date DATE,
        years_experience INT,
        monthly_salary NUMERIC(12,2),
        annual_bonus NUMERIC(12,2),
        performance_rating NUMERIC(3,2),
        projects_completed INT,
        leave_days_taken INT,
        overtime_hours NUMERIC(8,2),
        remote_worker BOOLEAN,
        employment_status VARCHAR(50),
        education_level VARCHAR(100),
        email VARCHAR(150),
        phone VARCHAR(30),
        emergency_contact VARCHAR(30),
        manager_name VARCHAR(100),
        certification VARCHAR(150),
        work_shift VARCHAR(50),
        performance_category VARCHAR(50),
        promotion_eligible BOOLEAN
    )
""")

conn.commit()
```

```python
for _, row in df.iterrows():

    cursor.execute("""
        INSERT INTO employees (
            employee_id,
            employee_code,
            first_name,
            last_name,
            gender,
            age,
            city,
            department,
            job_title,
            employment_type,
            join_date,
            years_experience,
            monthly_salary,
            annual_bonus,
            performance_rating,
            projects_completed,
            leave_days_taken,
            overtime_hours,
            remote_worker,
            employment_status,
            education_level,
            email,
            phone,
            emergency_contact,
            manager_name,
            certification,
            work_shift,
            performance_category,
            promotion_eligible
        )
        VALUES (
            %s, %s, %s, %s, %s, %s, %s, %s, %s, %s,
            %s, %s, %s, %s, %s, %s, %s, %s, %s, %s,
            %s, %s, %s, %s, %s, %s, %s, %s, %s
        )
    """, tuple(row))

conn.commit()

print("Data imported successfully!")
```

## Arithmatic Operator

```python
# 1. Display employee_id, monthly_salary, and monthly_salary + 10000 as increased_salary.
cursor.execute("select employee_id, monthly_salary, monthly_salary + 10000 as increased_salary from employees")
rows = cursor.fetchall()
print(rows)
```

```python
# 2. Calculate reduced_salary by subtracting 5000 from monthly_salary.
cursor.execute("select employee_id, monthly_salary, monthly_salary-5000 as reduced_salary from employees")
rows = cursor.fetchall()
print(rows)
```

## Mixed Challenge

```python
# 39. Find Active employees from Kathmandu or Lalitpur whose salary is between 90000 and 180000.
cursor.execute("""select employee_id, first_name, city, monthly_salary, employment_status from employees where 
                employment_status = 'Active' 
                and city in ('Kathmandu', 'Lalitput') 
                and monthly_salary between 90000 and 180000;
            """)

rows = cursor.fetchall()
print(rows)
```

```python
# 40. Find employees in IT or Analytics whose first_name starts with A and performance_rating is at least 4.
cursor.execute("""
    select employee_id, first_name, department, performance_rating from employees
    where department in ('IT', 'Analytics')
    and first_name like 'A%'
    and performance_rating >=4
    """)

rows = cursor.fetchall()
print(rows)
```

```python
# 41. Find employees who are not Interns and have completed more than 5 projects.
cursor.execute("""
    select employee_id, first_name, employment_type, projects_completed from employees
    where employment_type <> 'Intern'
    and projects_completed > 5;
    """)

rows = cursor.fetchall()
print(rows)
```

```python
# 42. Find employees with NULL certification OR NULL emergency_contact.
cursor.execute("""
    select employee_id, first_name, certification, emergency_contact from employees
    where certification is null 
    or emergency_contact is null
    """)

rows = cursor.fetchall()
print(rows)
```

```python
# 43. Find employees whose job_title contains Manager and whose employment_status is Active.
cursor.execute("""
    select employee_id, first_name, job_title, employment_status from employees
    where job_title like '%Manager%'
    and employment_status = 'Active'
    """)

rows = cursor.fetchall()
print(rows)
```

```python
# 44. Display employees aged between 30 and 50 who work remotely and have salary above 120000.
cursor.execute("""
    select employee_id, first_name, age, monthly_salary, remote_worker from employees
    where age between 30 and 50
    and remote_worker = 'Yes'
    and monthly_salary > 120000
    """)

rows = cursor.fetchall()
print(rows)
```

```python
# 45. Calculate annual_salary and total_compensation for employees in Finance, IT, and Analytics.
cursor.execute("""
    select employee_id, first_name, department, monthly_salary, annual_bonus, monthly_salary * 12 as annual_salary, (monthly_salary * 12) + annual_bonus as total_compensation from employees
    where department in ('IT', 'Analytics', 'Finance')
    """)

rows = cursor.fetchall()
print(rows)
```

```python
# 46. Find employees whose promotion_eligible = Yes AND performance_category = Excellent.
cursor.execute("""
    select employee_id, first_name, promotion_eligible, performance_category from employees
    where promotion_eligible = 'Yes'
    and performance_category = 'Excellent'
    """)

rows = cursor.fetchall()
print(rows)
```
