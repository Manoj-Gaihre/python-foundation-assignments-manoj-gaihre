### **Easy**

1. Build a list of the cubes of 1 through 10 using a list comprehension.
2. From `["  hi  ", "bye", "  yo"]`, produce a list of the strings stripped of surrounding whitespace.
3. Create a set of the unique vowels present in the string `"education is powerful"`.

````python
# 1. Build a list of the cubes of 1 through 10 using a list comprehension.
cube = [i**3 for i in range(1,11)]
print(cube)
````

````python
# 2. From `["  hi  ", "bye", "  yo"]`, produce a list of the strings stripped of surrounding whitespace.
raw_greeting = ["  hi  ", "bye", "  yo"]
cleaned_greeting = [word.strip() for word in raw_greeting]
print(cleaned_greeting)
````

````python
# 3. Create a set of the unique vowels present in the string `"education is powerful"`.
sentence = "education is powerful"
vowel = ['a','e','i','o','u']
contained_vowel = {i for i in sentence if i in vowel}
print(contained_vowel)
````

### **Medium**

1. Given `nums = [4, 7, 10, 13, 16, 19]`, use a dict comprehension to map each number to whether it's even (`{4: True, 7: False, ...}`).
2. From a list of words, build a dict mapping each word to its length, but only for words longer than 3 characters.
3. Use a generator expression to compute the sum of squares of all numbers from 1 to 100 — without ever building a list.

````python
''' 1. Given `nums = [4, 7, 10, 13, 16, 19]`, use a dict comprehension to map each number to whether 
it's even (`{4: True, 7: False, ...}`).'''

nums = [4, 7, 10, 13, 16, 19]
dict = {num:num%2==0 for num in nums}
print(dict)
````

````python
''' 2. From a list of words, build a dict mapping each word to its length, but only for words longer 
than 3 characters.
'''

words = ['hi','bye','hello','good','night','am']
dict = {word:len(word) for word in words if len(word)>3}
print(dict)
````

````python
''' 3. Use a generator expression to compute the sum of squares of all numbers from 1 to 100 — 
without ever building a list.'''

sqare = (i**2 for i in range(1,101))
print(sqare)

# if you want to see the result then we can look like this 
square = list(sqare)
print(square)
````

### **Hard**

1. Flatten a 2D list `matrix = [[1, 2], [3, 4], [5, 6]]` into a single list using a nested comprehension, keeping only even values.
2. Invert the dict `{"a": 1, "b": 2, "c": 2}` and explain (in a comment) what happens to the duplicate value `2`.
3. Given a list of log lines, build a dict that maps each log level (`"INFO"`, `"WARN"`, `"ERROR"`) to the **count** of lines containing it. (Hint: you may combine a comprehension with `str.count` or a helper.)

````python
''' 1. Flatten a 2D list `matrix = [[1, 2], [3, 4], [5, 6]]` into a single list using a nested 
comprehension, keeping only even values. '''

matrix = [[1, 2], [3, 4], [5, 6]]
print(matrix)

# converting matrix into list
martix_list = [x for row in matrix for x in row]
print(martix_list)

# list with only even values
martix_list_even = [x for row in matrix for x in row if x%2==0]
print(martix_list_even)
````

````python
''' 2. Invert the dict `{"a": 1, "b": 2, "c": 2}` and explain (in a comment) what happens to the 
duplicate value `2`.'''

dict = {"a": 1, "b": 2, "c": 2}
reversed_dict = {value:key for key,value in dict.items()}
print(reversed_dict)

'''here the duplicate value of 2 is not written for the both time because a dictionary cannot have 
duplicate keys and when the first 2 is encounter the value is written but when second value is encountered 
it simply overwrites over the previous one.'''
````

````python
''' 3. Given a list of log lines, build a dict that maps each log level (`"INFO"`, `"WARN"`, `"ERROR"`) 
to the **count** of lines containing it. (Hint: you may combine a comprehension with `str.count` or a 
helper.)'''

logs = [
    "INFO User logged in",
    "ERROR Database connection failed",
    "INFO User logged out",
    "WARN Low disk space",
    "ERROR File not found",
    "INFO Request completed"
]

levels = ["INFO", "WARN", "ERROR"]

# helper function
def count_level(level,log):
    return sum(level in line for line in log)

result = {level:count_level(level,logs) for level in levels}
print(result)
````

## Challenge Project

### Mini Log Analyzer

Build a small command-line tool that reads a server log and reports on it using comprehensions throughout.

**Spec**

Given a list of log lines like:

```text
2026-06-25 10:00 INFO user=alice action=login
2026-06-25 10:01 ERROR user=bob action=login msg=timeout
2026-06-25 10:02 WARN user=alice action=upload
2026-06-25 10:03 ERROR user=carol action=login msg=denied
```

Your tool should produce:

1. **Errors only** — a `list` of all lines containing `"ERROR"` (use a generator if the file is large).
2. **Unique users** — a `set` of every distinct `user=...` value mentioned.
3. **Level counts** — a `dict` mapping each level (`INFO`/`WARN`/`ERROR`) to how many lines have it.
4. **Login failures** — a `list` of `(user, msg)` pairs for `ERROR` lines whose `action=login`.


````python
path = 'server.log'

with open(path) as f:
    lines = f.read().splitlines()


# 1. Errors (lazy generator — only materialise when needed)
errors = (ln for ln in lines if "ERROR" in ln)

# 2. Unique users (set comprehension + a small parse helper)
def field(line, key):
    # returns the value of key=... or None
    for token in line.split():
        if token.startswith(key + "="):
            return token[len(key) + 1:]
    return None

users = {field(ln, "user") for ln in lines if field(ln, "user")}

# 3. Level counts (dict comprehension over the known levels)
levels = ["INFO", "WARN", "ERROR"]
counts = {lvl: sum(1 for ln in lines if lvl in ln) for lvl in levels}

# 4. Login failures — your turn: filter ERROR + action=login, collect (user, msg)
# Filter ERROR lines with action=login
# Then collect (user, msg) pairs
login_failures = [
    (field(ln, "user"), field(ln, "msg"))
    for ln in lines
    if "ERROR" in ln and "action=login" in ln
]


# Display the results

print("Errors:")
print(list(errors))

print("\nUnique users:")
print(users)

print("\nLevel counts:")
print(counts)

print("\nLogin failures:")
print(login_failures)
````
