```python
''' 1. Write a lambda `triple = lambda n: n * 3` and call it on `7`. Then rewrite it as a `def` and 
explain which PEP 8 prefers.'''
triple = lambda n: n * 3
print(triple(7))

# def version
def triple(n):
    return n*3

print(triple(7))

''' PEP 8 prefers the use of `def` instead of `lambda` because Lambdas are most useful when 
you need a small function temporarily and PEP 8 recommends using `def` for named functions, 
as it improves readability and maintainability of the code.'''
```

```python
''' 2. Given `nums = [5, 12, 17, 22, 30]`, use `map` to produce a list of each number squared, and 
`filter` to keep only those divisible by 3. Then rewrite both with comprehensions.'''
nums = [5, 12, 17, 22, 30]
squared = list(map(lambda x:x ** 2, nums))
print(f"map: {squared}")
filtered = list(filter(lambda x:x % 3 ==0, nums))
print(f"filter: {filtered}")

# Using comprehensions
print("Comprehensions:")
squared_comp = [x ** 2 for x in nums]
print(f"squared comp: {squared_comp}")
filterd_comp = [x for x in nums if x % 3 == 0]
print(f"filtered comp: {filterd_comp}")
```

1. Given `words = ['banana', 'kiwi', 'apple', 'fig']`, sort them (a) alphabetically, (b) by length, and (c) by length descending. Use `sorted` with `key=` for each.
2. You have `people = [{'name':'A','age':30}, {'name':'B','age':25}, {'name':'C','age':40}]`. Use `sorted` to order by age ascending, and `max`/`min` with `key=` to find the oldest and youngest person.
3. Build a **dispatch table** `ops = {'add': lambda a,b: a+b, 'mul': lambda a,b: a*b, 'sub': lambda a,b: a-b}` and write `calc(op, a, b)` that looks up and runs the right operation. Replace any `if/elif`.

```python
''' 1. Given `words = ['banana', 'kiwi', 'apple', 'fig']`, sort them (a) alphabetically, (b) by length, 
and (c) by length descending. Use `sorted` with `key=` for each.'''

words = ['banana', 'kiwi', 'apple', 'fig']

print("Alphabetically:")
print(sorted(words, key = lambda word: word))

print("By length:")
print(sorted(words, key=lambda word: len(word)))

print("By length descending:")
print(sorted(words, key = lambda word: len(word), reverse = True))
```

```python
''' 2. You have `people = [{'name':'A','age':30}, {'name':'B','age':25}, {'name':'C','age':40}]`. Use `sorted` 
to order by age ascending, and `max`/`min` with `key=` to find the oldest and youngest person.'''

people = [{'name':'A','age':30}, {'name':'B','age':25}, {'name':'C','age':40}]
print("Sorted by age ascending:")
print(sorted(people, key = lambda person: person["age"]))

print("Maximum age:")
print(max(people, key = lambda person:person['age']))

print("Minimum age:")
print(min(people, key = lambda person:person['age']))
```

```python
''' 3. Build a **dispatch table** `ops = {'add': lambda a,b: a+b, 'mul': lambda a,b: a*b, 'sub': lambda a,b: a-b}` 
and write `calc(op, a, b)` that looks up and runs the right operation. Replace any `if/elif`.'''

ops = {'add': lambda a,b: a+b, 'mul': lambda a,b: a*b, 'sub': lambda a,b: a-b}
def calc(op, a, b):
    if op not in ops:
        return f"Unknown Operation: {op}"

    else:
        return ops[op](a,b)


print(calc("add", 5,6))
print(calc("mul", 5,6))
print(calc("sub", 5,6))
```

1. Using `reduce`, compute the **product** of `[1,2,3,4,5]` (should be 120) and, separately, **flatten** `[[1,2],[3,4],[5]]` into `[1,2,3,4,5]`. Explain why `sum` can't do the product.

```python
''' 1. Using `reduce`, compute the **product** of `[1,2,3,4,5]` (should be 120) and, separately, 
**flatten** `[[1,2],[3,4],[5]]` into `[1,2,3,4,5]`. Explain why `sum` can't do the product.'''

from functools import reduce

numbers = [1,2,3,4,5]

product = reduce(lambda a,b : a * b , numbers)
print(product)

# flatten
nested = [[1,2],[3,4],[5]]
flattened = reduce(lambda a,b:a+b, nested)
print(flattened)

add = sum(flattened)
print(add)

# sum is specially designed for the addition not the multiplication.
```

```python
''' 2. Without using the built-in `max`, write `my_max(items, key=...)` using `reduce` that returns the item with the 
largest key value. Test it on the `people` list above to find the oldest.'''

from functools import reduce

people = [{'name':'A','age':30}, {'name':'B','age':25}, {'name':'C','age':40}]

def my_max(items, key):
    if not items:
        raise ValueError("my_max() is an empty sequence.")

    return reduce(
        lambda largest, current: current if key(current) > key(largest) else largest,
        items
    )

my_max(people, key = lambda x : x["age"])
```

### Mini Sales Data Pipeline (parse → map → filter → reduce)

Build a small data-processing pipeline over a list of sales records using only higher-order functions, `lambda`, `map`, `filter`, `reduce`, and `sorted`.

**Input data:**

```python
sales = [
    {"product": "pen",    "qty": 10, "price": 1.5},
    {"product": "laptop", "qty": 2,  "price": 800.0},
    {"product": "pencil", "qty": 50, "price": 0.5},
    {"product": "monitor","qty": 3,  "price": 150.0},
    {"product": "eraser", "qty": 25, "price": 0.2},
]
```

**Spec — produce:**

1. **map (transform):** a new list where each record gains a `"total"` key = `qty * price`.
2. **filter (select):** keep only records whose `total` is **≥ 50**.
3. **reduce (aggregate):** the **grand total revenue** of the filtered records.
4. **sorted (order):** the filtered records sorted by `total`, highest first.
5. Print a small report: each surviving product with its total, then the grand total.

**Hints:**

- For step 1, `map` a lambda that returns a new dict: `lambda r: {**r, "total": r["qty"]*r["price"]}`.
- For step 2, `filter(lambda r: r["total"] >= 50, ...)` — remember to `list()` if you'll reuse it.
- For step 3, `reduce(lambda acc, r: acc + r["total"], filtered, 0)` (note the `0` initial value — handy when the list might be empty).
- For step 4, `sorted(filtered, key=lambda r: r["total"], reverse=True)`.
- Don't name your accumulator `sum`. Use `total_revenue`.

**Stretch goals:**

- Replace the `map`/`filter` steps with comprehensions and compare readability.
- Build a **dispatch table** of report formatters (`{"short": ..., "verbose": ...}`) chosen by a `style` argument — the strategy pattern with lambdas.

```python
from functools import reduce

sales = [
    {"product": "pen",    "qty": 10, "price": 1.5},
    {"product": "laptop", "qty": 2,  "price": 800.0},
    {"product": "pencil", "qty": 50, "price": 0.5},
    {"product": "monitor","qty": 3,  "price": 150.0},
    {"product": "eraser", "qty": 25, "price": 0.2},
]

# map a new list where each record gains a total
mapped = list(map(lambda r: {**r, "total": r["qty"]*r["price"]} ,sales))

# filter: keep only records whose total is >= 50
filtered = list(filter(lambda r:r["total"]>=50, mapped))

# reduce: grand total revenue of filtered records
total_revenue = reduce(lambda acc, r: acc + r["total"], filtered, 0)

# sorted: filtered records sorted by highest total record first
sorted_sales = sorted(filtered, key = lambda x:x["total"], reverse=True)

# 5. REPORT
print("Sales Report")
print("-" * 30)

for record in sorted_sales:
    print(
        f"{record['product']}: "
        f"${record['total']:.2f}"
    )

print("-" * 30)
print(f"Grand Total Revenue: ${total_revenue:.2f}")
```

```python
sales_comp = [
    {"product": "pen",    "qty": 10, "price": 1.5},
    {"product": "laptop", "qty": 2,  "price": 800.0},
    {"product": "pencil", "qty": 50, "price": 0.5},
    {"product": "monitor","qty": 3,  "price": 150.0},
    {"product": "eraser", "qty": 25, "price": 0.2},
]

mapped_comp = [{**r, "total": r["qty"] * r["price"]} for r in sales_comp]
print(mapped_comp)

filtered_comp = [r for r in mapped_comp if r["total"]>=50]
print(filtered_comp)
```

```python
formatters = {
    "short": lambda r:
        f"{r['product']}: ${r['total']:.2f}",

    "verbose": lambda r:
        f"Product: {r['product']} | "
        f"Quantity: {r['qty']} | "
        f"Price: ${r['price']:.2f} | "
        f"Total: ${r['total']:.2f}"
}


def print_report(records, style="short"):
    if style not in formatters:
        print(f"Unknown style: {style}")
        return

    for record in records:
        print(formatters[style](record))


print_report(sorted_sales, "short")
```

```python
print_report(sorted_sales, "verbose")
```

### Alternative: Dict-of-lambdas Calculator

Build a calculator that dispatches operations from a dictionary of lambdas:

```python
ops = {
    "+": lambda a, b: a + b,
    "-": lambda a, b: a - b,
    "*": lambda a, b: a * b,
    "/": lambda a, b: a / b if b != 0 else "error: divide by zero",
}
```

**Spec:** write `calc(a, op, b)` that looks up `ops[op]` and calls it. Handle an unknown operator gracefully (for example, return a helpful message instead of `KeyError`). This is the **strategy pattern / dispatch table** in miniature — exactly how the AI tool-routing example earlier works.


```python
ops = {
    "+": lambda a, b: a + b,
    "-": lambda a, b: a - b,
    "*": lambda a, b: a * b,
    "/": lambda a, b: a / b if b != 0 else "error: divide by zero",
}

def calc(a, op, b):
    if op not in ops:
        return f"Unknown operation: {op}"

    return ops[op](a,b)

# Tests
print(calc(10, "+", 5))
print(calc(10, "-", 5))
print(calc(10, "*", 5))
print(calc(10, "/", 5))
print(calc(10, "/", 0))
print(calc(10, "%", 5))
```
