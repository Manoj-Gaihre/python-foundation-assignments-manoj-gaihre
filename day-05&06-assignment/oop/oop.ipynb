### **Easy**

1. Create a `Book` class with instance attributes `title` and `author`, and a method `describe()` that prints `"<title> by <author>"`. Instantiate two books and call `describe()` on each.
2. Add a class attribute `count` to `Book` that increments in `__init__` every time a book is created. Print the total after making three books.

```python
# 1. Create a `Book` class with instance attributes `title` and `author`, and a method `describe()` that prints `"<title> by <author>"`. Instantiate two books and call `describe()` on each.

class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def descibe(self):
        print(f"{self.title} is written by {self.author}")

book1 = Book("Ramayan","Balmiki")
book2 = Book("Mahabharat","Ved Vyasa")

book1.descibe()
book2.descibe()
```

```python
# 2. Add a class attribute `count` to `Book` that increments in `__init__` every time a book is created. Print the total after making three books.

class Book:
    count = 0
    def __init__(self, title, author):
        self.title = title
        self.author = author
        Book.count+=1



book1 = Book("Ramayan","Balmiki")
book2 = Book("Mahabharat","Ved Vyasa")
book2 = Book("Data Science", "David Lee")

print("Total count is: ", Book.count)    
```

### **Medium**

1. Build a `Shape` base class and `Circle`/`Rectangle` subclasses. Give each an `area()` method (override it). Store shapes in a list and print every area in a loop (polymorphism).
2. Create a `BankAccount` with a private `__balance`. Provide `deposit(amount)` and `withdraw(amount)` methods that validate (no negative deposits, no overdraft). Expose `get_balance()`.
3. Write a `@classmethod` alternative constructor `Book.from_string("title|author")` that splits the string and returns a `Book`.

```python
# 1. Build a `Shape` base class and `Circle`/`Rectangle` subclasses. Give each an `area()` method (override it). Store shapes in a list and print every area in a loop (polymorphism).

class Shape:
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

class Rectangle(Shape):
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def area(self):
        return self.length * self.breadth


shapes = [
    Circle(5),
    Rectangle(4,5),
    Circle(6)
]

for shape in shapes:
    print("Area: ", shape.area())
```

```python
# 2. Create a `BankAccount` with a private `__balance`. Provide `deposit(amount)` and `withdraw(amount)` methods that validate (no negative deposits, no overdraft). Expose `get_balance()`.

class BankAccount:
    def __init__(self, balance = 0):
        self.__balance = balance

    def deposit(self, amount):
        if amount < 0:
            raise ValueError("No negative deposits")

        self.__balance += amount


    def withdraw(self, amount):
        if amount > self.__balance:
            raise ValueError("No overdrafts. Insufficient balance")

        if amount < 0:
            raise ValueError("Withdraw amount cannot be negative.")

        self.__balance -= amount

    def get_balance(self):
        return self.__balance



account = BankAccount(1000)
print(account.get_balance())

account.deposit(1000)
print(account.get_balance())

account.withdraw(500)
print(account.get_balance())

account.withdraw(4000)


```

```python
# 3. Write a `@classmethod` alternative constructor `Book.from_string("title|author")` that splits the string and returns a `Book`.

class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    @classmethod
    def from_string(cls, text):
        title, author = text.split("|")
        return cls(title, author)

book = Book.from_string("Mahabharat|Ved Vyasa")

print(book.title)
print(book.author)
```

### **Hard**

1. Implement multilevel inheritance `Vehicle → Car → SportsCar`, each constructor using `super().__init__(...)`. The final object should carry attributes from all three levels.
2. Define an abstract `PaymentMethod(ABC)` with `@abstractmethod def pay(self, amount)`. Implement `CreditCard`, `Esewa`, and `MobileBanking`. Process a list of payments polymorphically. Verify that instantiating `PaymentMethod()` directly raises `TypeError`.

```python
# 1. Implement multilevel inheritance `Vehicle → Car → SportsCar`, each constructor using `super().__init__(...)`. The final object should carry attributes from all three levels.

class Vehicle():
    def __init__(self, brand):
        self.brand = brand

class Car(Vehicle):
    def __init__(self, brand, model):
        super().__init__(brand)
        self.model = model

class SportsCar(Car):
    def __init__(self, brand, model, top_speed):
        super().__init__(brand, model)
        self.top_speed = top_speed

car = SportsCar("Ferrari", "488", 340)

print(car.brand)
print(car.model)
print(car.top_speed)
```

```python
# 2. Define an abstract `PaymentMethod(ABC)` with `@abstractmethod def pay(self, amount)`. Implement `CreditCard`, `Esewa`, and `MobileBanking`. Process a list of payments polymorphically. Verify that instantiating `PaymentMethod()` directly raises `TypeError`.

from abc import ABC, abstractmethod

class PaymentMethod(ABC):
    @abstractmethod
    def pay(self, amount) : pass

class CreditCard(PaymentMethod):
    def pay(self, amount):
        print(f"Paid Rs. {amount} using credit card.")

class Esewa(PaymentMethod):
    def pay(self, amount):
        print(f"Paid Rs. {amount} using Esewa.")

class MobileBanking(PaymentMethod):
    def pay(self, amount):
        print(f"Paid Rs. {amount} using mobile banking.")

payments = [
    CreditCard(),
    Esewa(),
    MobileBanking()
]

for payment in payments:
    payment.pay(3000)

try:
    payment = PaymentMethod()
except TypeError as e:
    print("Error: ", e)
```

## Challenge Project — Mini Library Management System

Model a small library that exercises all four pillars.

**Spec**

- An abstract `LibraryItem(ABC)` with abstract method `describe()` and a `_checked_out` protected flag.
- Concrete subclasses `Book`, `DVD`, `Magazine` (inheritance), each overriding `describe()` (polymorphism).
- A `Member` class with a private `__borrowed_items` list and methods `borrow(item)` / `return_item(item)` with validation (encapsulation).
- A `Library` class holding items and members; method `checkout(member, item)` that flips `_checked_out` and updates the member.

**Hints**

- Use `from abc import ABC, abstractmethod` for `LibraryItem`.
- Override `describe()` differently per item type and loop over a mixed list to print all — that's polymorphism.
- Keep `__borrowed_items` private; expose a read-only view via a getter.
- Add a `@classmethod` `Library.from_catalog(list_of_dicts)` as an alternative constructor.
- Bonus: raise a clear error when a member borrows an already-checked-out item.

```python
from abc import ABC, abstractmethod

class LibraryItem(ABC):
    def __init__(self, title):
        self.title = title
        self._checked_out = False

    @abstractmethod
    def describe(): pass


class Book(LibraryItem):
    def __init__(self, title, author):
        super().__init__(title)
        self.author = author

    def describe(self):
        return f"Book: {self.title} by {self.author}"

class DVD(LibraryItem):
    def __init__(self, title, director):
        super().__init__(title)
        self.director = director

    def describe(self):
        return f"DVD: {self.title} directed by {self.director}"

class Magazine(LibraryItem):
    def __init__(self, title, issue):
        super().__init__(title)
        self.issue = issue

    def describe(self):
        return f"Magazine: {self.title}, Issue {self.issue}"


class Member:
    def __init__(self, name):
        self.name = name
        self.__borrowed_items = []

    def borrow(self, item):
        if item._checked_out:
            raise ValueError("Item already checked out  ")

        item._checked_out = True
        self.__borrowed_items.append(item)

    def return_item(self, item):
        if item not in self.__borrowed_items:
            raise ValueError("This item is not borrowed by this member.")

        item._checked_out = False
        self.__borrowed_items.remove(item)

    @property
    def borrowed_items(self):
        return tuple(self.__borrowed_items)




class Library:
    def __init__(self):
        self.items = []
        self.members = []

    def add_item(self, item):
        self.items.append(item)

    def add_member(self, member):
        self.members.append(member)

    def checkout(self, member, item):
        if member not in self.members:
            raise ValueError("Member is not registered in this library.")

        if item not in self.items:
            raise ValueError("Item does not belong to this library.")

        member.borrow(item)

    def return_item(self, member, item):
        if member not in self.members:
            raise ValueError("Member is not registered in this library.")

        member.return_item(item)

    @classmethod
    def from_catalog(cls, list_of_dicts):
        library = cls()

        for data in list_of_dicts:

            if data["type"] == "book":
                item = Book(data["title"], data["author"])

            elif data["type"] == "dvd":
                item = DVD(data["title"], data["director"])

            elif data["type"] == "magazine":
                item = Magazine(data["title"], data["issue"])

            else:
                raise ValueError("Unknown item type")

            library.add_item(item)

        return library

# Test the system
library = Library()

book = Book("Mahabharat", "Ved vyas")
dvd = DVD("Inception", "Christopher Nolan")
magazine = Magazine("Gorkha patra", 15)

library.add_item(book)
library.add_item(dvd)
library.add_item(magazine)

member = Member("Manoj")
library.add_member(member)

for item in library.items:
    print(item.describe())

print("\n")
# test checkout
library.checkout(member, book)
print("Borrowed items: ")

for item in member.borrowed_items:
    print(item.describe())

print("\n")

# try borrow the same book again
try:
    library.checkout(member, book)
except ValueError as e:
    print("Error:", e)


# Return item
library.return_item(member, book)

print("\nAfter returning:")
print("Book checked out:", book._checked_out)
print("Borrowed items:", [item.title for item in member.borrowed_items])


# Alternative constructor

catalog = [
    {
        "type": "book",
        "title": "Mahabharat",
        "author": "Ved Vyas"
    },
    {
        "type": "dvd",
        "title": "Inception",
        "director": "Christopher Nolan"
    },
    {
        "type": "magazine",
        "title": "Gorkha Patra",
        "issue": 15
    }
]

catalog_library = Library.from_catalog(catalog)

print("\nItems created from catalog:")
for item in catalog_library.items:
    print(item.describe())
```
